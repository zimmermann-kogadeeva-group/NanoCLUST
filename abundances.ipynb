```python
%matplotlib widget
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("output/mcd2022/Group1/combined.csv")
df["barcode"] = df["sample"].str.split("_", expand=True)[2]

df = (pd.DataFrame([(x[0][0], x[0][1], x[1]["rel_abundance"].mean()) 
                    for x in df.query("tax == 'S'").groupby(["barcode", "taxid"])], 
                   columns=["barcode", "species", "rel_abundance"])
       .pivot(index="barcode", columns="species", values="rel_abundance"))
df = df.div(df.sum(axis=1), axis=0)

fig, ax = plt.subplots(figsize=(8, 5))
fig.subplots_adjust(bottom=0.25, right=0.6)
df.plot(kind="bar", stacked=True, ax=ax)
ax.legend(loc="center", bbox_to_anchor=(1.4, 0.5))
ax.set_ylabel("Rel. abundance")
```
