```python
%matplotlib widget
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


df = pd.read_csv("output/combined.csv")
df["barcode"] = df["sample"].str.split("_", expand=True)[4]

df = (pd.DataFrame([(x[0][0], x[0][1], x[1]["rel_abundance"].mean()) 
                    for x in df.query("tax == 'S'").groupby(["barcode", "taxid"])], 
                   columns=["barcode", "species", "rel_abundance"])
       .pivot(index="barcode", columns="species", values="rel_abundance")
       .drop(columns=["267818", "33962", "483199"], index="unclassified"))
df = df.div(df.sum(axis=1), axis=0)

df_full = pd.read_csv("output/combined_full.csv")
df_full["barcode"] = df_full["sample"].str.split("_", expand=True)[4]

df_full = (pd.DataFrame([(x[0][0], x[0][1], x[1]["rel_abundance"].mean()) 
                         for x in df_full.query("tax == 'S'").groupby(["barcode", "taxid"])], 
                        columns=["barcode", "species", "rel_abundance"])
           .pivot(index="barcode", columns="species", values="rel_abundance")
           .drop(index="unclassified"))
df_full = df_full.div(df_full.sum(axis=1), axis=0)

fig, axs = plt.subplots(1, 2, figsize=(12, 5))
fig.subplots_adjust(bottom=0.22, right=0.75)

df.plot(kind="bar", stacked=True, ax=axs[0], legend=False)
df_full.plot(kind="bar", stacked=True, ax=axs[1])

axs[1].legend(loc="center", bbox_to_anchor=(1.4, 0.5))
axs[0].set_title("Partial fastq")
axs[1].set_title("Full fastq")
```
