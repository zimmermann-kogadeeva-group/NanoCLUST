```python
%matplotlib widget
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


df = pd.read_csv("output/combined.csv")
df["barcode"] = df["sample"].str.split("_", expand=True)[4]

df = (pd.DataFrame([(x[0][0], x[0][1], x[1]["rel_abundance"].mean()) 
                    for x in df.query("tax == 'S'").groupby(["barcode", "taxid"])], 
                   columns=["barcode", "species", "rel_abundance"])
       .pivot(index="barcode", columns="species", values="rel_abundance")
       .drop(columns=["267818", "33962", "483199"], index="unclassified"))
df = df.div(df.sum(axis=1), axis=0)

df2 = pd.read_csv("output/combined_full.csv")
df2["barcode"] = df2["sample"].str.split("_", expand=True)[4]

df2 = (pd.DataFrame([(x[0][0], x[0][1], x[1]["rel_abundance"].mean()) 
                         for x in df2.query("tax == 'S'").groupby(["barcode", "taxid"])], 
                        columns=["barcode", "species", "rel_abundance"])
           .pivot(index="barcode", columns="species", values="rel_abundance")
           .drop(index="unclassified"))
df2 = df2.div(df2.sum(axis=1), axis=0)

df3 = pd.read_csv("output/combined_full2.csv")
df3["barcode"] = df3["sample"].str.split("_", expand=True)[4]

df3 = (pd.DataFrame([(x[0][0], x[0][1], x[1]["rel_abundance"].mean()) 
                    for x in df3.query("tax == 'S'").groupby(["barcode", "taxid"])], 
                   columns=["barcode", "species", "rel_abundance"])
       .pivot(index="barcode", columns="species", values="rel_abundance"))
df3 = df3.div(df3.sum(axis=1), axis=0)

fig, axs = plt.subplots(1, 3, figsize=(12, 5))
fig.subplots_adjust(left=0.1, bottom=0.22, right=0.75)

df.plot(kind="bar", stacked=True, ax=axs[0], legend=False)
df2.plot(kind="bar", stacked=True, ax=axs[1], legend=False)
df3.plot(kind="bar", stacked=True, ax=axs[2])

axs[2].legend(loc="center", bbox_to_anchor=(1.65, 0.5))
axs[1].legend(loc="center", bbox_to_anchor=(2.8, 0.1))
labels = ("Partial fastq", "Full fastq", "Full combined fastq")
for ax, label in zip(axs, labels):
    ax.set_title(label)
```
